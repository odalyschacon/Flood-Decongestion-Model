# 3. Compare a Texas image pair and nearby rainfall

**Idea:** combine visual change with a related table of observations.
We compare the same Bear Creek footprint in a USDA baseline image and a
NOAA Harvey-response image, then look at the preceding two weeks of rain.


```python
from texas_flood import load_texas_images, show_images, ask_images, load_rainfall, plot_rainfall

pair = load_texas_images(["TX001_2016_NAIP", "TX001_2017_NOAA"])
pair["label"] = ["A: before", "B: response"]
show_images(pair)
```

## Ask about the change

`ask_images` sends both images in order. The dates are a year apart, so
season, vegetation and imaging differences can also explain changes.


```python
question = "Compare image A (first) and B (second). What visible changes in water coverage, roads and buildings do you see? Give two observations and one uncertainty."
print(ask_images(pair["path"], question))
```

## Add nearby rainfall

The helper follows the image ID to its assigned rain gauge and selects daily
observations before the image date. Missing or rejected readings remain missing.
Nearby rainfall provides context; it does not prove what is visible in the image.


```python
rain = load_rainfall("TX001_2017_NOAA", days_before=14)
plot_rainfall(rain)
```
